## I. Közös feladatok – együtt gépeljük és értelmezzük

Alapadatok: `nap = ['H', 'K', 'Sze', 'Cs', 'P', 'Szo']`, `terv = [118, 132, 125, 146, 137, 84]`, `kesz = [96, 107, 101, 119, 112, 68]`, `energia = [52, 57, 55, 61, 59, 38]`, `selejt = [8, 6, 11, 5, 9, 3]`.

```python
nap=['H','K','Sze','Cs','P',"Szo"]
terv=[118,132,125,146,137,84]
kesz=[96,107,101,119,112,68]
energia=[52,57,55,61,59,38]
selejt=[8,6,11,5,9,3]
```

**K1 ★ – Első vonal.** `fig, ax = plt.subplots()` után `ax.plot(nap, kesz, marker='o')`. Adj címet, „Nap” és „Elkészült (db)” tengelyfeliratot. Melyik a legnagyobb érték? **Ellenőrzés:** Cs, 119 db.

```python
fig,ax=plt.subplots()
ax.plot(nap,kesz,marker='o',color='#f9d18b',linewidth=2)
ax.set(title="Napi elkészült mennyiségek",xlabel="Nap",ylabel="Elkészült (db)")
fig.tight_layout()
```

**K2 ★ – Első oszlop.** Külön ábrán `ax.bar(nap, energia)`. A függőleges tengely „Energia (kWh)” legyen. Melyik a legkisebb napi fogyasztás? **Ellenőrzés:** Szo, 38 kWh.

```python
fig,ax=plt.subplots()
ax.bar(nap,energia,color="#f59e0b")
ax.set(title="Napi energiafogyasztás",xlabel="Nap",ylabel="Energia (kWh)")
```

**K3 ★ – Terv és tény.** Egy ábrán két vonal, mindkét `plot` kapjon `label` értéket, utána `ax.legend()`. Melyik nap volt a legnagyobb abszolút elmaradás? **Ellenőrzés:** Cs, 27 db.

```python
fig,ax=plt.subplots()
ax.plot(nap,terv,marker="o",label="Tervezet")
ax.plot(nap,kesz,marker="s",label="Elkészült")
ax.set(title="Tervezett és elkészült darabszám",xlabel="Nap",ylabel="Mennyiség (db)")
ax.legend()
ax.grid(alpha=0.25)
```

**I1 ★ – Selejt naponta.** Ábrázold a `selejt` listát. Válassz oszlopot vagy vonalat. Indokold egy mondatban a választást. A tengely mértékegysége db.


```python
fig,ax=plt.subplots()
ax.bar(nap,selejt,color="#ef4444")
ax.set(title="Napi selejtmennyiség", xlabel="Nap",ylabel="Selejt (db)")
```

```python
#csv_beolvas -> adatokat olvass be CSV fileból
import csv
from pathlib import Path

ADATMAPPA=Path('.')
def csv_beolvas(fajlnev):
  with open(ADATMAPPA/fajlnev,encoding='utf-8-sig',newline='') as f:
    return list(csv.DictReader(f,delimiter=';'))
```

**I2 ★ – Terv és tény a második adatkészleten.** A `masodik_het.csv` fájlból készíts kétvonalas grafikont az elkészült és tervezett darabokról. CSV-ben a számokat `int`-té kell alakítani. Melyik napon volt a legnagyobb elmaradás? **Ellenőrzés:** Szerda, 15 db.

```python
masodik=csv_beolvas('masodik_het.csv')
print(masodik)
nap2=[s['nap'] for s in masodik]
terv2=[int(s['tervezett_db']) for s in masodik]
kesz2=[int(s['elkeszult_db']) for s in masodik]

print(nap2)
print(terv2)
print(kesz2)

fig, ax = plt.subplots()
ax.plot(nap2, terv2, marker='o', label='Tervezett')
ax.plot(nap2, kesz2, marker='s', label='Elkészült')
ax.set(title='Második hét - terv és tény', xlabel='Nap', ylabel='Mennyiség (db)')
ax.legend()
```

**I3 – Két hét összehasonlítása.** A két hét elkészült darabjait ábrázold egy ábrán, azonos napok szerint. Az első hét a fenti lista, a második hét a CSV. Az egyik vonal szaggatott legyen (`linestyle='--'`). Mely napokon nőtt a termelés a második héten? **Ellenőrzés:** H, K, Sze, Cs, P; szombaton csökkent.

```python
fig,ax=plt.subplots()
ax.plot(nap2,kesz, marker='o', label='Első hét')
ax.plot(nap2,kesz2, marker='s', linestyle='--', label='Második hét')
ax.set(title="Elkészült darabszám két héten",xlabel="Nap",ylabel="Elkészült (db)")
ax.legend()
```

**I4 – Különbség kiszámítása.** Listakomprehenzióval számítsd a `terv[i] - kesz[i]` eltérést mind a hat napra. Ábrázold oszlopon. Miért beszélünk elmaradásról, és nem százalékról? **Ellenőrzés:** `[22, 25, 24, 27, 25, 16]` db.

```python
elmaradas=[terv[i] - kesz[i] for i in range(len(nap))]
print(elmaradas)
fig,ax=plt.subplots()
ax.bar(nap,elmaradas,color='#FFA500')
ax.set(title="Napi elmaradás a tervhez képst",xlabel='Nap',ylabel='Elmaradás (db)')
```

**I5 – Selejtarány.** `100 * selejt[i] / kesz[i]` segítségével készíts százalékos oszlopdiagramot. Most az y felirata `%` legyen, ne db. Melyik napon a legmagasabb az arány? **Ellenőrzés:** Szerda, körülbelül 10,9%.

```python
selejtarany = [100 * selejt[i] / kesz[i] for i in range(len(nap))]
fig, ax = plt.subplots()
ax.bar(nap, selejtarany, color='#dc2626')
ax.set(title='Napi selejtarány', xlabel='Nap', ylabel='Selejtarány (%)')
```

**I6 – Üzemóra és energia.** A `muhely_adatok.csv` fájlból készíts két külön oszlopdiagramot üzemórára és napi energiára. Miért kell a szombati napi kWh-t óvatosan értelmezni? **Ellenőrzés:** szombaton 12 üzemóra szerepel.

```python
muhely = csv_beolvas('muhely_adatok.csv')
m_nap = [s['nap'] for s in muhely]
uzemora = [int(s['uzemora']) for s in muhely]
m_energia = [int(s['energia_kwh']) for s in muhely]

fig, ax = plt.subplots()
ax.bar(m_nap, uzemora, color='#0ea5e9')
ax.set(title='Napi üzemidő', xlabel='Nap', ylabel='Üzemidő (óra)')


fig, ax = plt.subplots()
ax.bar(m_nap, m_energia, color='#f59e0b')
ax.set(title='Napi energiafogyasztás', xlabel='Nap', ylabel='Energia (kWh)')
```

**I7 – Két ábra egy fig-en.** `fig, axes = plt.subplots(1, 2, figsize=(12, 4))`: bal oldalon elkészült db, jobb oldalon selejt db. Mindkét tengelynek külön cím és y felirat kell. Ez a feladat két eltérő skálát őriz meg.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(nap, kesz, color='#2563eb')
axes[0].set(title='Elkészült mennyiség', xlabel='Nap', ylabel='Elkészült (db)')
axes[1].bar(nap, selejt, color='#ef4444')
axes[1].set(title='Selejtmennyiség', xlabel='Nap', ylabel='Selejt (db)')
fig.suptitle('Termelés és minőség – külön skálán')
fig.tight_layout()
plt.show()
```

**Ö1 ★ – Termelés a műhelyben.** `muhely_adatok.csv`: terv és elkészült db két vonalon, címmel, jelmagyarázattal. Mutasd meg a legnagyobb elmaradást. **Ellenőrzés:** Cs, 27 db.

```python
m_terv = [int(s['tervezett_db']) for s in muhely]
m_kesz = [int(s['elkeszult_db']) for s in muhely]
m_elmaradas = [m_terv[i] - m_kesz[i] for i in range(len(m_nap))]
idx = m_elmaradas.index(max(m_elmaradas))

fig, ax = plt.subplots()
ax.plot(m_nap, m_terv, marker='o', label='Tervezett')
ax.plot(m_nap, m_kesz, marker='s', label='Elkészült')
ax.annotate(f'{m_elmaradas[idx]} db elmaradás', xy=(m_nap[idx], m_kesz[idx]),
            xytext=(m_nap[idx], m_kesz[idx] - 30),
            arrowprops={'arrowstyle': '->'}, ha='center')
ax.set(title='Műhely termelése – terv és tény', xlabel='Nap', ylabel='Mennyiség (db)')
ax.legend()
fig.tight_layout()
fig.savefig('o1_termeles.png', dpi=160)
plt.show()
```

**Ö2 ★ – Működés vagy minőség.** Ugyanabból a CSV-ből külön oszlopdiagram a napi energiáról **vagy** a selejtről. Írd a tengelyre a kWh **vagy** db egységet. Indokold, miért nem a termeléssel közös y tengelyen ábrázolod.

```python
fig, ax = plt.subplots()
ax.bar(m_nap, m_energia, color='#f59e0b')
ax.set(title='Műhely napi energiafogyasztása', xlabel='Nap', ylabel='Energia (kWh)')
fig.tight_layout()
fig.savefig('o2_energia.png', dpi=160)
plt.show()
```

**Ö3 – Termelés egységnyi energiára.** `elkeszult_db / energia_kwh` naponta, y egység db/kWh. Melyik nap a legmagasabb? **Ellenőrzés:** csütörtök, kb. 1,95 db/kWh. Az arányszám nem az okok bizonyítéka.

```python
db_per_kwh = [m_kesz[i] / m_energia[i] for i in range(len(m_nap))]
idx = db_per_kwh.index(max(db_per_kwh))
fig, ax = plt.subplots()
ax.bar(m_nap, db_per_kwh, color='#10b981')
ax.set(title='Termelés egységnyi energiára', xlabel='Nap', ylabel='Termelés (db/kWh)')
fig.tight_layout()
fig.savefig('o3_db_per_kwh.png', dpi=160)
plt.show()
print(f'Legmagasabb: {m_nap[idx]}, {db_per_kwh[idx]:.2f} db/kWh')
```

**Ö4 – Tervteljesítés.** `100 * elkeszult_db / tervezett_db` százalékban. Készíts oszlopdiagramot, és jelöld a legmagasabb napot. **Ellenőrzés:** péntek, kb. 81,8%.

```python
teljesites = [100 * m_kesz[i] / m_terv[i] for i in range(len(m_nap))]
idx = teljesites.index(max(teljesites))
szinek = ['#2563eb'] * len(m_nap)
szinek[idx] = '#16a34a'
fig, ax = plt.subplots()
ax.bar(m_nap, teljesites, color=szinek)
ax.annotate(f'{teljesites[idx]:.1f}%', xy=(m_nap[idx], teljesites[idx]),
            xytext=(m_nap[idx], teljesites[idx] + 4), ha='center',
            arrowprops={'arrowstyle': '->'})
ax.set(title='Napi tervteljesítés', xlabel='Nap', ylabel='Tervteljesítés (%)')
ax.set_ylim(0, 100)
fig.tight_layout()
plt.show()
print(f'Legmagasabb: {m_nap[idx]}, {teljesites[idx]:.1f}%')
```

**Ö5 – Második heti termelés.** `masodik_het.csv`: új vonaldiagram két adatsorral, a legnagyobb elmaradás megnevezése. **Ellenőrzés:** szerda, 15 db.

```python
fig, ax = plt.subplots()
ax.plot(nap2, terv2, marker='o', label='Tervezett')
ax.plot(nap2, kesz2, marker='s', label='Elkészült')
ax.set(title='Második heti termelés', xlabel='Nap', ylabel='Mennyiség (db)')
ax.legend()
fig.tight_layout()
plt.show()
print('Legnagyobb elmaradás: Sze, 15 db')
```

**Ö6 – Hétközi átlag.** A két hét hétfő–pénteki elkészült darabszámának átlagát ábrázold két oszlopon. Szombatot mindkét hétből hagyd ki. **Ellenőrzés:** első hét 107 db/nap, második 113,2 db/nap.

```python
atlag1 = sum(kesz[:5]) / 5
atlag2 = sum(kesz2[:5]) / 5
fig, ax = plt.subplots()
bars = ax.bar(['Első hét', 'Második hét'], [atlag1, atlag2], color=['#2563eb', '#7c3aed'])
ax.bar_label(bars, fmt='%.1f db/nap')
ax.set(title='Hétfő–pénteki átlagos termelés', xlabel='Hét', ylabel='Átlag (db/nap)')
ax.set_ylim(0, 125)
fig.tight_layout()
plt.show()
print(f'Első hét: {atlag1:.1f} db/nap; második hét: {atlag2:.1f} db/nap')
```

**Ö7 – Selejt két héten.** A két hét selejtdarabszámát vonalon vagy csoportos oszlopokon ábrázold nap szerint. Figyelj az azonos db egységre és a jelmagyarázatra. Melyik napon nagyobb az első heti selejt? **Ellenőrzés:** hétfő, szerda, péntek.

```python
selejt2 = [int(s['selejt_db']) for s in masodik]
fig, ax = plt.subplots()
ax.plot(nap, selejt, marker='o', label='Első hét')
ax.plot(nap, selejt2, marker='s', linestyle='--', label='Második hét')
ax.set(title='Selejtmennyiség két héten', xlabel='Nap', ylabel='Selejt (db)')
ax.legend()
fig.tight_layout()
plt.show()
nagyobb = [nap[i] for i in range(len(nap)) if selejt[i] > selejt2[i]]
print('Az első heti selejt ezeken a napokon nagyobb:', ', '.join(nagyobb))
```

**Ö8 – Hőmérséklet idősora.** `gepadatok.csv`: a gép hőmérsékletét vonallal ábrázold az időpontok sorrendjében, °C egységgel. Mikor a legmagasabb? **Ellenőrzés:** 12:00, 46 °C.

```python
gep = csv_beolvas('gepadatok.csv')
ido = [s['idopont'] for s in gep]
homerseklet = [int(s['homerseklet_c']) for s in gep]
fig, ax = plt.subplots()
ax.plot(ido, homerseklet, marker='o', color='#dc2626')
ax.set(title='A gép hőmérséklete', xlabel='Időpont', ylabel='Hőmérséklet (°C)')
fig.tight_layout()
plt.show()
idx = homerseklet.index(max(homerseklet))
print(f'Legmagasabb: {ido[idx]}, {homerseklet[idx]} °C')
```

**Ö9 – Vibráció összehasonlítása.** `gepadatok.csv`: vibráció az időpontok szerint, mm/s egységgel. Ábrázold oszlopon, és nevezd meg a legmagasabb mérést. **Ellenőrzés:** 12:00, 4,8 mm/s. Az adatokhoz nem adtunk biztonsági határértéket, ezért ne minősítsd a mérést veszélyesnek.

```python
vibracio = [float(s['vibracio_mm_s'].replace(',', '.')) for s in gep]
fig, ax = plt.subplots()
ax.bar(ido, vibracio, color='#8b5cf6')
ax.set(title='A gép vibrációja', xlabel='Időpont', ylabel='Vibráció (mm/s)')
fig.tight_layout()
plt.show()
idx = vibracio.index(max(vibracio))
print(f'Legmagasabb mérés: {ido[idx]}, {vibracio[idx]:.1f} mm/s')
```

**Ö10 – Gépterhelés és hőmérséklet.** Két külön tengelyen, ugyanabban a fig-ben ábrázold a terhelést (%) és a hőmérsékletet (°C) időpont szerint. Fogalmazz meg egy megfigyelést, és külön mondd el, miért nem bizonyítja ez az ok-okozatot.

```python
terheles = [int(s['terheles_szazalek']) for s in gep]
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(ido, terheles, marker='o', color='#2563eb')
axes[0].set(title='Gépterhelés', xlabel='Időpont', ylabel='Terhelés (%)')
axes[1].plot(ido, homerseklet, marker='s', color='#dc2626')
axes[1].set(title='Hőmérséklet', xlabel='Időpont', ylabel='Hőmérséklet (°C)')
fig.suptitle('Gépterhelés és hőmérséklet külön tengelyeken')
fig.tight_layout()
plt.show()
```

**Ö11 – Szűrés.** Az Ö8 ábrát készítsd el újra csak a 08:00–14:00 közötti mérésekkel. Hány mérés maradt? **Ellenőrzés:** 4.

```python
szurt = [s for s in gep if '08:00' <= s['idopont'] <= '14:00']
szurt_ido = [s['idopont'] for s in szurt]
szurt_hom = [int(s['homerseklet_c']) for s in szurt]
fig, ax = plt.subplots()
ax.plot(szurt_ido, szurt_hom, marker='o', color='#dc2626')
ax.set(title='Géphőmérséklet 08:00–14:00 között', xlabel='Időpont', ylabel='Hőmérséklet (°C)')
fig.tight_layout()
plt.show()
print(f'Megmaradt mérések száma: {len(szurt)}')
```

**Ö12 – Saját kérdés.** Válassz egy megválaszolható kérdést a három CSV valamelyikéből. Készíts indokolt diagramot, pontos egységgel, és írd le a választ a konkrét értékekkel. A választott kérdés különbözzön az Ö1–Ö11 témájától.

```python
kwh_per_ora = [m_energia[i] / uzemora[i] for i in range(len(m_nap))]
fig, ax = plt.subplots()
ax.bar(m_nap, kwh_per_ora, color='#0f766e')
ax.set(title='Óránkénti energiafelhasználás', xlabel='Nap', ylabel='Energiaintenzitás (kWh/óra)')
fig.tight_layout()
plt.show()
idx_max = kwh_per_ora.index(max(kwh_per_ora))
idx_min = kwh_per_ora.index(min(kwh_per_ora))
print(f'Legmagasabb: {m_nap[idx_max]}, {kwh_per_ora[idx_max]:.2f} kWh/óra')
print(f'Legalacsonyabb: {m_nap[idx_min]}, {kwh_per_ora[idx_min]:.2f} kWh/óra')
```
